### Listas

```python
### Listas
minha_lista = ["vinicios", "lucas", "juan", "jean", True, False, 1, 2, 3, 4, 5]
print(minha_lista[0]) # Retorna o primeiro elemento da lista
print(minha_lista[-1]) # Retorna o último elemento da lista
print(minha_lista[0:5]) # Retorna os elementos do índice 0 ao 4
print(minha_lista[:5]) # Retorna os elementos do início da lista até o índice 4
print(minha_lista[2:]) # Retorna os elementos do índice 2 até o final da lista
```

```python
# Operações com listas
minha_lista = ["vinicios", "lucas", "juan", "jean", True, False, 1, 2, 3, 4, 5]
minha_lista.append("novo elemento") # Adiciona um elemento no final da lista 
minha_lista
```

```python
minha_lista.remove("novo elemento") # Remove um elemento da lista
minha_lista
```

```python
minha_lista[-1] = "mudanças para melhor"
minha_lista
```

```python
indice = minha_lista.index("vinicios")
print(f"O indice do elemento mais importante da lista é: {indice}")
```

```python
minha_lista.insert(-3, "novo elemento") # Adiciona um elemento no início da lista
minha_lista
```

```python
# Método pop
minha_lista.pop(-2) # Remove o elemento correspondente ao índice da lista
minha_lista
```

### Loops 

```python
### If e else simples para validar se a variável corresponde aos valores que queremos que ela tenha. No primeiro caso, se positivo, ele retorna verdadeiro, caso contrário, falso.
nome = input("Por favor, digite seu nome: ")

if nome in ["joão", "vinicios", "lucas", "juan", "jean"]:
    print("Verdadeiro")
else:
    print("Falso")
```
